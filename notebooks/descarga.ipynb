Objetivo:

Descargar los datos de la EPH desde 2016 a 2026 tanto de hogares como de individuos.

Problema: como los datos "crudos" tienen demasiadas columnas que no vamos a usar y eso ocuparía mucha memoria voy a descargarlos por separado para hacer la limpieza luego.

pyeph genera una carpeta "pyeph" con los zip de cada trimestre, los descargo en la carpeta data/raw para luego limpiar uno por uno antes de unificar.

```python
import os
import pandas as pd
import pyeph
import shutil
from datetime import datetime
```

```python



# Configuración de rangos
ano_inicio = 2016
ano_actual = datetime.now().year
tipos_base = ['individual', 'hogar']

# Estructura para almacenar los datos descargados
bases = {'individual': [], 'hogar': []}

for ano in range(ano_inicio, ano_actual + 1):
    for periodo in range(1, 5):  # Trimestres 1 a 4
        for tipo in tipos_base:
            try:
                # Descarga la base EPH
                pyeph.obtener(
                    data='eph',
                    ano=ano,
                    periodo=periodo,
                    tipo_base=tipo
                )
                print(f"✔ Descargado: {tipo.capitalize()} {ano} - Trimestre {periodo}")
            except Exception:
                # Se salta trimestres no publicados o inexistentes
                print(f"✖ No disponible: {tipo.capitalize()} {ano} - Trimestre {periodo}")
```

```python
origen = "pyeph" if os.path.exists("pyeph") else os.path.join("..", "pyeph")

# Ruta destino: data/raw/pyeph
destino = os.path.join("..", "data", "raw", "pyeph")

if os.path.exists(origen):
    # Si la carpeta destino ya existe, la elimina previamente para evitar el error de "Directorio no vacío"
    if os.path.exists(destino):
        shutil.rmtree(destino)
        
    # Mueve la carpeta completa con sus archivos ZIP/DB
    shutil.move(origen, destino)
    print(f"✔ Carpeta movida exitosamente a: {destino}")
else:
    print(f"✖ No se encontró la carpeta '{origen}'. Verifica que exista en el directorio actual.")
```
